```python
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
transactions = {
    "Transaction_ID": [101,102,103,104,105,106,107,108,109,110],
    "Customer": [
        "Rahul","Priya","Amit","Sneha","Arjun",
        "Kiran","Meera","Rohan","Anjali","Vikram"
    ],
    "Amount": [
        500,1200,50000,700,150000,
        900,450,250000,1000,80000
    ],
    "Location": [
        "Chennai","Mumbai","Delhi","Chennai","Delhi",
        "Mumbai","Chennai","Delhi","Mumbai","Delhi"
    ]
}

df = pd.DataFrame(transactions)

df
```

```python
# Transactions greater than ₹50,000 are marked as Fraud

df["Fraud"] = df["Amount"].apply(lambda x: "Fraud" if x > 50000 else "Normal")

df
```

```python
df["Fraud"].value_counts()
```

```python
fraud_count = df["Fraud"].value_counts()

plt.figure(figsize=(6,4))
fraud_count.plot(kind="bar")

plt.title("Fraud vs Normal Transactions")
plt.xlabel("Transaction Type")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6,6))

plt.pie(
    fraud_count,
    labels=fraud_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Fraud vs Normal Transactions")

plt.show()
```

```python
fraud_transactions = df[df["Fraud"] == "Fraud"]

fraud_transactions
```
